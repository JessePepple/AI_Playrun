```python
from openai import OpenAI
import os
from dotenv import load_dotenv

if os.environ.get("API_KEY") is None:
    raise ValueError("API_KEY environment variable is not set. Please set it in your .env file.")
else:
    load_dotenv()  # Load environment variables from .env file
```

```python
from langchain_openai import ChatOpenAI
load_dotenv()

llm = ChatOpenAI(model= "gpt-3.5-turbo", temperature=0, api_key= os.environ.get("API_KEY"))

response  = llm.invoke("Translate the English sentence to Spanish: Hello, my friend")
response.content
```
